## Pregunta No. 2

```python
ticketsNorte = df_ventas.loc[df_ventas["sucursal"] == "Norte", "id_ticket"].nunique()
ticketsSur = df_ventas.loc[df_ventas["sucursal"] == "Sur", "id_ticket"].nunique()
ticketsCentro = df_ventas.loc[df_ventas["sucursal"] == "Central", "id_ticket"].nunique()

def masRegistro():
    if ticketsNorte > ticketsCentro and ticketsNorte > ticketsSur:
        print("La sucursal Norte vendió más.")
    elif ticketsSur > ticketsNorte and ticketsSur > ticketsCentro:
        print("La sucursal Sur vendió más.")
    elif ticketsCentro > ticketsNorte and ticketsCentro > ticketsSur:
        print("La sucursal Centro vendió más.")

tabla2 = pd.DataFrame({
    "Sucursales" : ["Norte", "Sur", "Central"],
    "Cantidad de tickets": [ticketsNorte, ticketsSur, ticketsCentro],
})
display(tabla2)
masRegistro()
```

## Pregunta 4

```python
df_productos = pd.read_csv('productos_.csv', dtype={'producto': str})

escritura = df_productos[df_productos["categoria"] == "Escritura"]["precio_lista"]
archivo = df_productos[df_productos["categoria"] == "Archivo"]["precio_lista"]
accesorios = df_productos[df_productos["categoria"] == "Accesorios"]["precio_lista"]
papelYcuadernos = df_productos[df_productos["categoria"] == "Papel y cuadernos"]["precio_lista"]

tabla4 = pd.DataFrame({
    "Categoría": ["Escritura", "Archivo", "Accesorios", "Papel y Cuadernos"],
    "Promedio de costos": [escritura.mean(), archivo.mean(), accesorios.mean(), papelYcuadernos.mean()],
    "Costo máximo": [escritura.max(), archivo.max(), accesorios.max(), papelYcuadernos.max()],
    "Costo mínimo": [escritura.min(), archivo.min(), accesorios.min(), papelYcuadernos.min()]
})
display(tabla4)
```

## Pregunta 5

```python
# Unimos ventas con productos
df_completo = pd.merge(df_ventas,df_productos[["id_producto", "categoria"]],on="id_producto",how="left")

# Creamos la tabla cruzada
tabla5 = pd.crosstab(
    index=df_completo["sucursal"],
    columns=df_completo["categoria"],
    values=df_completo["cantidad"],
    aggfunc="sum",
    margins=True,
    margins_name="Total"
)

display(tabla5)

# Comprobamos los totales
totalVentas = df_ventas["cantidad"].sum()
totalTabla = tabla5.loc["Total", "Total"]

print("Total de unidades en ventas:", totalVentas)
print("Total de unidades en tabla:", totalTabla)

if totalVentas == totalTabla:
    print("Los totales coinciden.")
else:
    print("Los totales no coinciden.")
```

## Pregunta 6

```python
df_ventas["total_ventas"] = df_ventas["cantidad"] * df_ventas["precio_unitario"]

for sucursal in df_ventas["sucursal"].unique():
    ventasSucursal = df_ventas.loc[df_ventas["sucursal"] == sucursal, "total_ventas"].sum()

tabla6 = df_ventas.groupby("sucursal")["total_ventas"].sum().reset_index()

tabla6["total_ventas"] = tabla6["total_ventas"].round(2)

display(tabla6)
```

## Pregunta 7

```python
ticketsUnidades = df_ventas.groupby("id_ticket")["cantidad"].sum()

tickets7U = ticketsUnidades[ticketsUnidades <= 7].count()
tickets8o14U = ticketsUnidades[(ticketsUnidades >= 8) & (ticketsUnidades <= 14)].count()
ticketsMas = ticketsUnidades[ticketsUnidades >= 15].count()

tabla7 = pd.DataFrame({
    "Unidades por tickets": ["7U", "8 a 14 U", "Más de 15 U"],
    "Cantidad de tickets": [tickets7U, tickets8o14U, ticketsMas],
})
display(tabla7)
```

## Pregunta 8

```python
df_ventas["fecha"] = pd.to_datetime(df_ventas["fecha"])

df_ventas["mes"] = df_ventas["fecha"].dt.to_period("M")

df_ventas["total_venta"] = (df_ventas["cantidad"] * df_ventas["precio_unitario"])

tabla8 = df_ventas.groupby("mes").agg(Tickets=("id_ticket", "nunique"),Importe_Total=("total_venta", "sum"))

display(tabla8)

mesMayor = tabla8["Importe_Total"].idxmax()

print("Mes con mayor importe:", mesMayor)
print("Importe:", tabla8.loc[mesMayor, "Importe_Total"])
```

## Pregunta No. 9

```python
ventasProducto = df_ventas.groupby(
    ["sucursal", "id_producto"],
    as_index=False
)["cantidad"].sum()

ventasProducto["maximo"] = ventasProducto.groupby(
    "sucursal"
)["cantidad"].transform("max")


masVendidos = ventasProducto[ventasProducto["cantidad"] == ventasProducto["maximo"]]

tabla9 = pd.merge(
    masVendidos,
    df_productos[["id_producto", "producto"]],
    on="id_producto",
    how="left"
)

tabla9 = tabla9[
    ["sucursal", "id_producto", "producto", "cantidad"]
]

display(tabla9)
```

## Pregunta 10

```python
df_clientes = pd.read_csv('clientes.csv', dtype={'nombre':str})

df_completo = pd.merge(
    df_ventas,
    df_clientes,
    left_on="id_cliente",
    right_on="id",
    how="left"
)

df_completo["importe"] = (df_completo["cantidad"] * df_completo["precio_unitario"])

tabla10 = df_completo.groupby("tipo_cliente").agg(
    Clientes_Distintos=("id_cliente", "nunique"),
    Tickets=("id_ticket", "nunique"),
    Importe_Total=("importe", "sum")
)

tabla10["Importe_Total"] = tabla10["Importe_Total"].round(2)

display(tabla10)
```
